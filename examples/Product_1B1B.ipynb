```python
import qcombo
```

```python
# Helper function: display SymPy expressions as LaTeX in Jupyter
from IPython.display import display, Latex
def jupyterDisplay(expr, title=None):
    """
    Display SymPy expression in LaTeX format in Jupyter Notebook
    """
    if expr == 0 or expr is None:
        display(Latex(f"$$0$$"))
        return
    latex_expr = qcombo.texExp(expr)
    if title:
        print(title)
    display(Latex(f"$${latex_expr}$$"))
```

```python
left  = [["a"],['b']]
right = [["k"],["l"]]

operator_1B1B = qcombo.bodys(left,right)

operator_1B1B.parallel = False
operator_1B1B.show_process = False
```

```python
operator_1B1B.commutator()
product_1B1B = operator_1B1B.gw
jupyterDisplay(product_1B1B)
```

The product of NO1B and NO1B operator is the same as formula in Ref.*H. Hergert*,**In-Medium Similarity Renormalization Group for Closed and Open-Shell Nuclei**, Appendix (A.2)

$$:A_b^a :: A_l^k := :A_{bl}^{ak}: - \lambda_l^a :A_b^k: - \xi_b^k :A_l^a: - \lambda_l^a \xi_b^k + \lambda_{bl}^{ak}$$

```python
# trans xi to (lambda - delta)
operator_1B1B.applyRule("xi")
commutator_1B1B = operator_1B1B.cmtRule
jupyterDisplay(commutator_1B1B)
```

The commutator of NO1B and NO1B operator is the same as formula in Ref.*H. Hergert*,**In-Medium Similarity Renormalization Group for Closed and Open-Shell Nuclei**, Appendix (A.5)

$$[:A_b^a:, :A_l^k:] = \delta_b^k :A_l^a: - \delta_l^a :A_b^k: + \lambda_l^a \delta_b^k - \lambda_b^k \delta_l^a$$
