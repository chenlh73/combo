This module is intended to test whether parallel computing produces the same results as non-parallel computing

```python
import qcombo
```

```python
# display sympy expression in latex in jupternotebook
from IPython.display import display, Latex
def jupyterDisplay(expr):
    latex_expr = qcombo.texExp(expr)
    display(Latex(f"$${latex_expr}$$"))
```

```python
# Define the indices
left = [['a','b'],['c','d']]
right = [['i','j'],['k','l']]

# Create two bodys objects with the same parameters except parallel
operators_parallel = qcombo.bodys(left, right)
operators_parallel.wick_mode = 'MR'
operators_parallel.show_process = True
operators_parallel.parallel = True

operators_non_parallel = qcombo.bodys(left, right)
operators_non_parallel.wick_mode = 'MR'
operators_non_parallel.show_process = True
operators_non_parallel.parallel = False
```

Test 1: Check whether the generalized Wick theorem results are consistent

```python
# Calculate the commutator
operators_parallel.commutator()
operators_non_parallel.commutator()

# Compare the generalized Wick theorem results
print("parallel gw == non-parallel gw:", operators_parallel.gw == operators_non_parallel.gw)
# Compare the commutator results
print("parallel cmt == non-parallel cmt:", operators_parallel.cmt == operators_non_parallel.cmt)
```

Test 2: Check whether the applyRule results are consistent

```python
# Apply simplification rules
operators_parallel.applyRule('both')
operators_non_parallel.applyRule('both')

# Compare the results after applying rules
print("parallel cmtRule == non-parallel cmtRule:", operators_parallel.cmtRule == operators_non_parallel.cmtRule)
```

Test 3: Check whether the regulate results are consistent

```python
# Filter and regulate terms
operators_parallel.regulate(filterbody=0)
operators_non_parallel.regulate(filterbody=0)

# Compare filtered terms
print("parallel filterTerms == non-parallel filterTerms:", operators_parallel.filterTerms == operators_non_parallel.filterTerms)
# Compare canonicalized results
print("parallel canon == non-parallel canon:", operators_parallel.canon == operators_non_parallel.canon)
```

Test 4: Check whether the simplifyUseBoth results are consistent

```python
# Simplify the expressions
simplified_parallel = qcombo.simplifyUseBoth(operators_parallel.canon,parallel=True)
simplified_non_parallel = qcombo.simplifyUseBoth(operators_non_parallel.canon,parallel=False)

print("parallel simplified == non-parallel simplified:", simplified_parallel == simplified_non_parallel)
```

Test 7: Check the filterLambda function with parallel vs non-parallel

```python
# filter lambda
lambdabody = 1

filterLambda_parallel = qcombo.filterLambdaBody(simplified_parallel,lambdabody,parallel=True)
filterLambda_non_parallel = qcombo.filterLambdaBody(simplified_non_parallel,lambdabody,parallel=False)

print("parallel filterLambda == non-parallel filterLambda:", filterLambda_parallel == filterLambda_non_parallel)
```

Summary: If all tests print True, parallel computing produces consistent results with non-parallel computing.
