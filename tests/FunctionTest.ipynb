This module is intended to test whether the modified qcombo functions produce the same output as the original ones

```python
import qcombo
```

```python
# display sympy expression in latex in jupternotebook
from IPython.display import display, Latex
def jupyterDisplay(expr):
    # texExp is function to trans Sympy expression to latex
    latex_expr = qcombo.texExp(expr)
    # display latex in jupternotebook
    display(Latex(f"$${latex_expr}$$"))
```

```python
#define the indices

# left=[['a_1'],['a_2']]
left=[['a','b'],['c','d']]
right=[['i','j'],['k','l']]
# right=[['i','j','k'],['l','m','n']]
#right=[['i'],['j']]

operators = qcombo.bodys(left,right)

# Set control parameters
operators.wick_mode = 'MR'
operators.show_process = True
operators.parallel = True
```

```python
all_indices = qcombo.flatten_indices(operators.indices)

print(type(all_indices))
```

```python
#calculate the commutator
operators.commutator()
#display the Latex of product of left and right operator 
#based on the generalized wick theorem
jupyterDisplay(operators.gw)
#display the Latex of commutator of left and right operator 
#based on the generalized wick theorem
jupyterDisplay(operators.cmt)
```

```python
#diagonalize the one-body density matrix
operators.applyRule('both')
#display the commutator after Diagonalizing
jupyterDisplay(operators.cmtRule)
```

Below we check the xiRule and natRule functions

```python
from qcombo import SimplifyRule

expr = operators.cmt
# qcombo.jupyterDisplay(expr)
all_indices = qcombo.flatten_indices(operators.indices)

#
expr_oold = SimplifyRule.xiRule_oold(expr)
expr_old = SimplifyRule.xiRule_old(expr)
expr_new = SimplifyRule.xiRule(expr,all_indices)

print("oold = new :", expr_oold == expr_new)
print("old = new :", expr_old == expr_new)

#
expr_oold_2 = SimplifyRule.natRule_oold(expr_oold)
expr_old_2 = SimplifyRule.natRule_old(expr_old)
expr_new_2 = SimplifyRule.natRule(expr_new,all_indices)

print("oold = new :", expr_oold_2 == expr_new_2)
print("old = new :", expr_old_2 == expr_new_2)
```

```python
#filter the desired terms and normalize the indices
operators.regulate(filterbody=0)
#filtered terms with Matrix elements
jupyterDisplay(operators.filterTerms)
#contract the delta indices and canonicalize indices
jupyterDisplay(operators.canon)
```

Below we check the filterbody function

```python
from qcombo import Filter

expr = operators.allTerms
body = 0

#
expr_1 = Filter.filterbody_old(expr,body,show_process=True)
expr_2 = Filter.filterbody(expr,body,show_process=True,parallel=True)

print("old = new: ",expr_1 == expr_2)
```

Below we check the tools.indicesMultToSimp function

```python
from qcombo import canonical,tools

canonTemp=canonical.canonicalize(operators.filterTerms, parallel=True, show_process=True)
jupyterDisplay(canonTemp)

canon,indicesSet=tools.indicesMultToSimp_old(canonTemp)
jupyterDisplay(canon)

canon_new,indicesSet_new=tools.indicesMultToSimp(canonTemp)
jupyterDisplay(canon_new)

print("old = new:",canon_new == canon)
print("old = new:",indicesSet_new == indicesSet)
```
