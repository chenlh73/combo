```python
import qcombo
```

```python
# Helper function: display SymPy expressions as LaTeX in Jupyter
from IPython.display import display, Latex
def jupyterDisplay(expr, title=None):
    """
    Display SymPy expression in LaTeX format in Jupyter Notebook
    """
    if expr == 0 or expr is None:
        display(Latex(f"$$0$$"))
        return
    latex_expr = qcombo.texExp(expr)
    if title:
        print(title)
    display(Latex(f"$${latex_expr}$$"))
```

```python
# operator 1
left_1 = [['a_1'],["a_2"]]
right_1  = [[r"\alpha"],[r"\beta"]]

operator_1 = qcombo.bodys(left_1,right_1)
operator_1.parallel = False
operator_1.show_process = False

# operator 2
left_2  = [[r"\gamma"],[r"\Omega"]]
right_2 = [['b_1'],["b_2"]]

operator_2 = qcombo.bodys(left_2,right_2)

operator_2.parallel = False
operator_2.show_process = False
```

## Before regualator step, the indices is not changed.

```python
operator_1.commutator()
operator_2.commutator()

print("product of indices 1:")
jupyterDisplay(operator_1.gw)
print("product of indices 2:")
jupyterDisplay(operator_2.gw)
```

```python
print("commutator of indices 1:")
jupyterDisplay(operator_1.cmt)
print("commutator of indices 2:")
jupyterDisplay(operator_2.cmt)
```

```python
operator_1.applyRule()
operator_2.applyRule()
print("Diagonalized commutator of indices 1:")
jupyterDisplay(operator_1.cmtRule)
print("Diagonalized commutator of indices 2:")
jupyterDisplay(operator_2.cmtRule)
```

## After regulate step, the all the indices are ordered from "a" to "z", so the final expressions are the same.

```python

operator_1.regulate(filterbody=1)
operator_2.regulate(filterbody=1)

print("Canonicalized commutator of indices 1:")
jupyterDisplay(operator_1.canon)
print("Canonicalized commutator of indices 2:")
jupyterDisplay(operator_2.canon)
```
